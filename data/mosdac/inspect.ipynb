```python
import xarray as xr

ds = xr.open_dataset("E06OCML4AC_20240101_25km_v1.0.0.nc")

print("DIMENSIONS")
for k, v in ds.sizes.items():
    print(f"  {k}: {v}")

print("\nVARIABLES")
for name, var in ds.data_vars.items():
    print(f"\n{name}")
    print(f"  dims: {var.dims}")
    print(f"  shape: {var.shape}")
    print(f"  dtype: {var.dtype}")
    print(f"  units: {var.attrs.get('units')}")
    print(f"  long_name: {var.attrs.get('long_name')}")

print("\nCOORDINATES")
for name, var in ds.coords.items():
    print(f"  {name}: {var.dims}, {var.shape}")

print("\nGLOBAL ATTRIBUTES")
for k, v in ds.attrs.items():
    print(f"  {k}: {v}")
```

```python
import xarray as xr

ds = xr.open_dataset("E06OCML4AC_20240101_25km_v1.0.0.nc")

print("LAT:")
print("min =", float(ds.lat.min()))
print("max =", float(ds.lat.max()))
print("sample =", ds.lat.values[:5])

print("\nLON:")
print("min =", float(ds.lon.min()))
print("max =", float(ds.lon.max()))
print("sample =", ds.lon.values[:5])

print("\nTIME:")
print(ds.time.values)

print("\nCHLA:")
print(
    "min =", float(ds.chla.min(skipna=True)),
    "max =", float(ds.chla.max(skipna=True)),
    "mean =", float(ds.chla.mean(skipna=True))
)
```
